**ASSIGNMENT Ordinal/label encoding**

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    "Satisfaction": ["Poor", "Good", "Average", "Excellent"],
    "Membership": ["Bronze", "Silver", "Gold", "Bronze"],
    "Purchase_Amount": [500, 1200, 800, 2500]
})

encoder = OrdinalEncoder(categories=[
    ["Poor", "Average", "Good", "Excellent"],
    ["Bronze", "Silver", "Gold"]
])

df[["Satisfaction", "Membership"]] = encoder.fit_transform(
    df[["Satisfaction", "Membership"]]
)

print(df)
```

```python
print(df.groupby("Satisfaction")["Purchase_Amount"].mean())

print(df.groupby("Satisfaction")["Purchase_Amount"].mean())
```

```python
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder

df = pd.DataFrame({
    "Study_Level": ["Low", "Medium", "High", "Low"],
    "City": ["Bhopal", "Delhi", "Mumbai", "Delhi"],
    "Result": ["Fail", "Pass", "Pass", "Fail"]
})

df["Study_Level"] = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
).fit_transform(df[["Study_Level"]])

df["Result"] = LabelEncoder().fit_transform(df["Result"])

df = pd.get_dummies(df, columns=["City"], dtype=int)

print(df)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X = df.drop("Result", axis=1)
y = df["Result"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=42
)

model = DecisionTreeClassifier()
model.fit(X_train, y_train)

pred = model.predict(X_test)

print("Predictions:", pred)
print("Accuracy:", accuracy_score(y_test, pred))
```

```python
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    "Size": ["Small", "Medium", "Large", "XL"],
    "Color": ["Red", "Blue", "Green", "Red"],
    "Satisfaction": ["Poor", "Good", "Average", "Excellent"]
})

encoder = OrdinalEncoder(categories=[
    ["Small", "Medium", "Large", "XL"],
    ["Poor", "Average", "Good", "Excellent"]
])

df[["Size", "Satisfaction"]] = encoder.fit_transform(
    df[["Size", "Satisfaction"]]
)

print(df)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

ct = ColumnTransformer([
    ("ord", OrdinalEncoder(categories=[
        ["Small", "Medium", "Large", "XL"],
        ["Poor", "Average", "Good", "Excellent"]
    ]), ["Size", "Satisfaction"]),
    ("nom", OneHotEncoder(), ["Color"])
])

result = ct.fit_transform(pd.DataFrame({
    "Size": ["Small", "Medium", "Large", "XL"],
    "Color": ["Red", "Blue", "Green", "Red"],
    "Satisfaction": ["Poor", "Good", "Average", "Excellent"]
}))

print(result)
```
