```python
salaries = [18, 20, 22, 22, 24, 25, 25, 26, 28, 30, 32, 35, 38, 42, 150]

def calculate_mean(data):
    return sum(data) / len(data)


def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)

    if n % 2 == 1:
        return sorted_data[n // 2]
    else:
        return (sorted_data[n // 2 - 1] + sorted_data[n // 2]) / 2


def calculate_modes(data):
    frequency = {}

    for value in data:
        frequency[value] = frequency.get(value, 0) + 1

    max_frequency = max(frequency.values())

    if max_frequency == 1:
        return []

    return [value for value, count in frequency.items()
            if count == max_frequency]


print("Mean:", calculate_mean(salaries))
print("Median:", calculate_median(salaries))
print("Mode:", calculate_modes(salaries))

```

```python
new_salaries = [salary for salary in salaries if salary != 150]

old_mean = calculate_mean(salaries)
old_median = calculate_median(salaries)

new_mean = calculate_mean(new_salaries)
new_median = calculate_median(new_salaries)

new_modes = calculate_modes(new_salaries)

print("New Mean:", new_mean)
print("New Median:", new_median)
print("New Mode:", new_modes)

mean_change = ((new_mean - old_mean) / old_mean) * 100
median_change = ((new_median - old_median) / old_median) * 100

print("Percentage change in mean:", round(mean_change, 2), "%")
print("Percentage change in median:", round(median_change, 2), "%")
```

```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("tips")

def calculate_mean(data):
    return sum(data) / len(data)


def calculate_median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 1:
        return data[n // 2]
    else:
        return (data[n // 2 - 1] + data[n // 2]) / 2

def calculate_modes(data):
    frequency = {}

    for value in data:
        frequency[value] = frequency.get(value, 0) + 1

    max_frequency = max(frequency.values())

    return [
        value for value, count in frequency.items()
        if count == max_frequency
    ]

for column in ["total_bill", "tip"]:
    values = df[column].tolist()

    print(f"\n{column}")
    print("Mean:", round(calculate_mean(values), 2))
    print("Median:", round(calculate_median(values), 2))
    print("Mode(s):", calculate_modes(values))


for meal in ["Lunch", "Dinner"]:
    bills = df.loc[df["time"] == meal, "total_bill"].tolist()

    print(f"\n{meal}")
    print("Mean bill:", round(calculate_mean(bills), 2))
    print("Median bill:", round(calculate_median(bills), 2))

```

```python

data_1 = [12, 15, 15, 18, 20, 22, 25]
data_2 = [5, 5, 10, 10, 15, 20]
data_3 = [7, 14, 21, 28, 35]
data_4 = [100, 100, 100, 100]
data_5 = [1, 2, 3, 4, 5, 100]


def calculate_mean(data):
    return sum(data) / len(data)



def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)

    if n % 2 == 1:
        return sorted_data[n // 2]
    else:
        middle1 = sorted_data[n // 2 - 1]
        middle2 = sorted_data[n // 2]
        return (middle1 + middle2) / 2



def calculate_mode(data):
    frequency = {}

    for value in data:
        frequency[value] = frequency.get(value, 0) + 1

    max_frequency = max(frequency.values())

    if max_frequency == 1:
        return []

    modes = [
        value for value, count in frequency.items()
        if count == max_frequency
    ]

    return modes



datasets = {
    "Data 1": data_1,
    "Data 2": data_2,
    "Data 3": data_3,
    "Data 4": data_4,
    "Data 5": data_5
}


print("=" * 65)
print("         CENTRAL TENDENCY CALCULATOR REPORT")
print("=" * 65)

print(f"{'Dataset':<12}{'Mean':<12}{'Median':<12}{'Mode'}")
print("-" * 65)

for name, data in datasets.items():
    mean = calculate_mean(data)
    median = calculate_median(data)
    mode = calculate_mode(data)

    mode_result = mode if mode else "No mode"

    print(f"{name:<12}{mean:<12.2f}{median:<12.2f}{mode_result}")

print("=" * 65)
```

```python

orders = [
    1, 1, 1, 2, 2, 2, 2, 3, 3, 4,
    4, 5, 5, 6, 8, 10, 15, 20, 25, 80
]


def calculate_mean(data):
    return sum(data) / len(data)


def calculate_median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 1:
        return data[n // 2]
    else:
        return (data[n // 2 - 1] + data[n // 2]) / 2

def calculate_modes(data):
    frequency = {}

    for value in data:
        frequency[value] = frequency.get(value, 0) + 1

    max_frequency = max(frequency.values())

    if max_frequency == 1:
        return []

    return [
        value for value, count in frequency.items()
        if count == max_frequency
    ]

print("Original Dataset")
print("Mean:", round(calculate_mean(orders), 2))
print("Median:", calculate_median(orders))
print("Mode:", calculate_modes(orders))

new_orders = orders.copy()
new_orders.remove(max(new_orders))

print("\nAfter Removing Largest Observation")
print("Mean:", round(calculate_mean(new_orders), 2))
print("Median:", calculate_median(new_orders))
print("Mode:", calculate_modes(new_orders))

```

```python
import pandas as pd

sales = [
    12, 15, 14, 16, None,
    15, 18, 14, 17, 16,
    None, 19, 15, 120, 14
]

s = pd.Series(sales)

print("Sales Series:")
print(s)

print("\nMissing values:")
print(s.isnull())

missing_count = s.isnull().sum()
total_count = len(s)
missing_percentage = (missing_count / total_count) * 100

print("\nNumber of missing values:", missing_count)
print("Percentage of missing values:", missing_percentage, "%")

clean_sales = s.dropna()

print("\nStatistics excluding missing values:")
print("Mean:", round(clean_sales.mean(), 2))
print("Median:", clean_sales.median())
print("Mode(s):", clean_sales.mode().tolist())

Q1 = clean_sales.quantile(0.25)
Q3 = clean_sales.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)


outliers = clean_sales[
    (clean_sales < lower_bound) |
    (clean_sales > upper_bound)
]

print("Potential outliers:", outliers.tolist())

```

```python
original_sales = s.dropna()


sales_without_outlier = original_sales[
    original_sales != 120
]


original_mean = original_sales.mean()
original_median = original_sales.median()
original_modes = original_sales.mode().tolist()


new_mean = sales_without_outlier.mean()
new_median = sales_without_outlier.median()
new_modes = sales_without_outlier.mode().tolist()


mean_change = (
    (new_mean - original_mean) / original_mean
) * 100

median_change = (
    (new_median - original_median) / original_median
) * 100

comparison = pd.DataFrame({
    "Measure": ["Mean", "Median", "Mode"],
    "Before Exclusion": [
        original_mean,
        original_median,
        original_modes
    ],
    "After Exclusion": [
        new_mean,
        new_median,
        new_modes
    ]
})

print(comparison)

print("\nPercentage change in mean:",
      round(mean_change, 2), "%")

print("Percentage change in median:",
      round(median_change, 2), "%")
```
