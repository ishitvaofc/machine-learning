```python
scores = [45, 50, 52, 55, 58, 60, 65, 75]

n = len(scores)
mean = sum(scores) / n

total = sum((x - mean) ** 2 for x in scores)

pv = total / n
sv = total / (n - 1)

print("Mean:", mean)
print("Population Variance:", pv)
print("Sample Variance:", sv)
print("Population Standard Deviation:", pv ** 0.5)
print("Sample Standard Deviation:", sv ** 0.5)
```

Population variance: Divides the sum of squared differences by n.
Sample variance: Divides the sum of squared differences by n - 1.

```python
import seaborn as sns

df = sns.load_dataset('tips')

for col in ['total_bill', 'tip']:
    data = df[col]
    mean = data.mean()
    variance = ((data - mean) ** 2).mean()
    std = variance ** 0.5
    cv = (std / mean) * 100

    print("\n", col)
    print("Mean:", round(mean, 2))
    print("Population Variance:", round(variance, 2))
    print("Population Standard Deviation:", round(std, 2))
    print("CV:", round(cv, 2), "%")
```

```python
largest = df.nlargest(5, 'total_bill')
print("Largest bills:")
print(largest['total_bill'].tolist())


df2 = df.drop(largest.index)

for col in ['total_bill', 'tip']:
    old = df[col]
    new = df2[col]

    print("\n", col)
    print("Original Mean:", round(old.mean(), 2))
    print("New Mean:", round(new.mean(), 2))

    old_var = ((old - old.mean()) ** 2).mean()
    new_var = ((new - new.mean()) ** 2).mean()

    print("Original Variance:", round(old_var, 2))
    print("New Variance:", round(new_var, 2))
    print("Original SD:", round(old_var ** 0.5, 2))
    print("New SD:", round(new_var ** 0.5, 2))
```

```python
import math

data_1 = [10, 20, 30, 40, 50]
data_2 = [20, 20, 20, 20, 20]
data_3 = [5, 10, 15, 20, 100]
data_4 = [12, 14, 16, 18, 20, 22]

def mean(data):
    return sum(data) / len(data)

def pop_var(data):
    m = mean(data)
    return sum((x - m) ** 2 for x in data) / len(data)

def sample_var(data):
    m = mean(data)
    return sum((x - m) ** 2 for x in data) / (len(data) - 1)

def pop_std(data):
    return math.sqrt(pop_var(data))

def sample_std(data):
    return math.sqrt(sample_var(data))

datasets = [data_1, data_2, data_3, data_4]

for i, data in enumerate(datasets, 1):
    print(f"\nData {i}: {data}")
    print("Mean:", round(mean(data), 2))
    print("Population Variance:", round(pop_var(data), 2))
    print("Sample Variance:", round(sample_var(data), 2))
    print("Population SD:", round(pop_std(data), 2))
    print("Sample SD:", round(sample_std(data), 2))
```

```python
measurements = [12, 15, 18, 20, 25, 30, 40]

data2 = [x + 10 for x in measurements]
data3 = [x * 4 for x in measurements]

def calculate(data):
    mean = sum(data) / len(data)
    variance = sum((x - mean) ** 2 for x in data) / len(data)
    sd = variance ** 0.5
    return mean, variance, sd

for data in [measurements, data2, data3]:
    print("Data:", data)
    print("Mean, Variance, SD:", 
          [round(x, 2) for x in calculate(data)])
```

```python
data4 = [x * -2 for x in measurements]

mean, variance, sd = calculate(data4)

print("Data:", data4)
print("Mean:", round(mean, 2))
print("Variance:", round(variance, 2))
print("Standard Deviation:", round(sd, 2))
```

```python
import numpy as np

sales = [12, 14, 15, 15, 16, 17, 18, 19, 20, 21, 22, 24, 25, 120]

mean = np.mean(sales)
median = np.median(sales)
variance = np.var(sales)
sd = np.std(sales)

Q1 = np.percentile(sales, 25)
Q3 = np.percentile(sales, 75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = [x for x in sales if x < lower or x > upper]

print("Mean:", round(mean, 2))
print("Median:", round(median, 2))
print("Population Variance:", round(variance, 2))
print("Population SD:", round(sd, 2))
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
print("Outliers:", outliers)
```

```python
clean = [x for x in sales if x not in outliers]

mean2 = np.mean(clean)
median2 = np.median(clean)
variance2 = np.var(clean)
sd2 = np.std(clean)

print("New Mean:", round(mean2, 2))
print("New Median:", round(median2, 2))
print("New Variance:", round(variance2, 2))
print("New SD:", round(sd2, 2))

print("Variance decrease (%):",
      round((variance - variance2) / variance * 100, 2))

print("SD decrease (%):",
      round((sd - sd2) / sd * 100, 2))
```
