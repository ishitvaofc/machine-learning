**ASSIGNMENT OF ROW\COLUMN\FEATURE\TARGET**

```python

import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112],
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Vikas", "Pooja",
             "Karan", "Simran", "Arjun", "Meena", "Rohit", "Anjali"],
    "Age": [19, 20, 18, 21, 20, 19, 22, 18, 21, 20, 19, 22],
    "Hours_Studied": [2, 5, 3, 8, 6, 4, 9, 2, 7, 5, 6, 8],
    "Attendance": [65, 85, 70, 95, 88, 75, 98, 60, 90, 82, 86, 94],
    "Marks": [48, 78, 55, 92, 84, 67, 96, 45, 89, 76, 81, 93]
}

df = pd.DataFrame(data)

print("First five records:")
print(df.head())

print("\nLast three records:") 
print(df.tail(3))
```

```python

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("DataFrame dimensions:", df.shape)

# Verify expected dimensions
if df.shape == (12, 6):
    print("Dimensions verified successfully!")
else:
    print("Dimensions do not match.")
```

```python

columns = df.columns.tolist()

print("Column names:", columns)

identifiers = ["Student_ID"]
categorical_attributes = ["Name"]
numerical_attributes = ["Age", "Hours_Studied", "Attendance", "Marks"]

print("Identifiers:", identifiers)
print("Categorical attributes:", categorical_attributes)
print("Numerical attributes:", numerical_attributes)
```

```python

selected_data = df[["Name", "Hours_Studied", "Marks"]]

print("Selected columns:")
print(selected_data)


average_marks = df["Marks"].mean()

print("\nAverage marks:", round(average_marks, 2))

above_average = selected_data[selected_data["Marks"] > average_marks]

print("\nStudents scoring above average:")
print(above_average)
```

```python

print("Original indices:")
print(above_average.index.tolist())

reset_data = above_average.reset_index(drop=True)

print("\nDataFrame after resetting the index:")
print(reset_data)

print("\nNew indices:")
print(reset_data.index.tolist())

```
