```python
import numpy as np
from sklearn.preprocessing import PowerTransformer
```

```python
X = np.array([
    [-1],
    [0],
    [3],
    [5],
    [10],
    [50]
])

pt = PowerTransformer(method="yeo-johnson")

X_transformed = pt.fit_transform(X)

print("original:")
print(X.ravel())

print("\ntransformed:")
print(np.round(X_transformed.ravel(), 3))

print("\nlearned lambda:")
print(pt.lambdas_)
```

```python
import numpy as np
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry'],
    'Age': [25, np.nan, 35, 45, np.nan, 29, 41, 38],                   # Numerical with nulls
    'Salary': [50000, 62000, np.nan, 85000, 72000, 48000, np.nan, 67000], # Numerical with nulls
    'Experience_Years': [2, 5, 10, np.nan, 8, 3, 15, 7],             # Numerical with nulls
    'Department': ['IT', 'HR', 'Finance', 'IT', np.nan, 'Marketing', 'Finance', 'HR'], # Categorical with nulls
    'Education': ['Bachelors', 'Masters', 'PhD', 'Bachelors', 'Masters', np.nan, 'Bachelors', 'PhD'], # Categorical with nulls
    'City': ['New York', 'London', 'Paris', np.nan, 'Berlin', 'New York', 'London', 'Paris'], # Categorical with nulls
    'Purchased': ['Yes', 'No', 'No', 'Yes', 'Yes', 'No', np.nan, 'Yes'] # Target (Categorical)
}

df = pd.DataFrame(data)
print(df)
print("\n--- Missing Value Counts ---")
print(df.isnull().sum())
```
